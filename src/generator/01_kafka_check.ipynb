```python
## thiskafka connect test code is from databricvks genie

import confluent_kafka
from databricks.sdk.runtime import dbutils
from confluent_kafka.admin import AdminClient

print("confluent-kafka version:", confluent_kafka.version())

conf = {
    "bootstrap.servers": dbutils.secrets.get(scope="pa_clock_scope", key="kafka-bootstrap"),
    "security.protocol": "SASL_SSL",
    "sasl.mechanisms": "PLAIN",
    "sasl.username": dbutils.secrets.get(scope="pa_clock_scope", key="kafka-api-key"),
    "sasl.password": dbutils.secrets.get(scope="pa_clock_scope", key="kafka-api-secret"),
}
md = AdminClient(conf).list_topics(timeout=10)
print("Connected. Topics:", sorted(t for t in md.topics))
```

```python
## thiskafka connect test code is from claude


from confluent_kafka.admin import AdminClient

SCOPE = "pa_clock_scope"
conf = {
    "bootstrap.servers": dbutils.secrets.get(SCOPE, "kafka-bootstrap"),
    "security.protocol": "SASL_SSL",
    "sasl.mechanisms": "PLAIN",
    "sasl.username": dbutils.secrets.get(SCOPE, "kafka-api-key"),
    "sasl.password": dbutils.secrets.get(SCOPE, "kafka-api-secret"),
}

md = AdminClient(conf).list_topics(timeout=15)
print("Brokers:", len(md.brokers))
print("Topics:", sorted(md.topics))
assert "pa-events" in md.topics, "Topic pa-events not found"
print("OK: credentials, network and topic all good")
```
